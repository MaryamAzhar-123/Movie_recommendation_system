__Item Based Recommendation System__

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
user_movie_matrix_filled = user_movie_matrix.fillna(0)
```

__Transpose the matrix__

```python
movie_similarity = cosine_similarity(user_movie_matrix_filled.T)
```

```python
movie_similarity.shape
```

__Recommend Movie Function__

```python
def recommend_movies(title , n=5):
    find_title = df1[df1["Title"] == title]['Movie_id'].values[0]
    address_in_matrix = user_movie_matrix_filled.columns.get_loc(find_title)

    similarity_score = movie_similarity[address_in_matrix]
    sorting = similarity_score.argsort()[::-1]

    top = sorting[1:6]

    Recommendation = user_movie_matrix_filled.columns[top]
    Recommended_Movies = df1[df1['Movie_id'].isin(Recommendation)]

    return Recommended_Movies['Title']
```

__Train test split__

```python
from sklearn.model_selection import train_test_split
```

```python
train,test = train_test_split(df, test_size=0.2, random_state=42)
```

```python
print("Train: ",train.shape)
print("Test: ",test.shape)
```

```python
train_matrix = train.pivot_table(
    index = 'User_id',
    columns = 'Movie_id',
    values = 'rating'
)
print("Train Matrix Shape: ",train_matrix.shape)
```

__Train the Model on SVD__

```python
train_matrix_filled = train_matrix.fillna(0)
evaluation = TruncatedSVD(
    n_components=50,
    random_state=42
)

train_transformed = evaluation.fit_transform(train_matrix_filled)
print("Transformed Shape: ", train_transformed.shape)
```

```python
predict_rating_evaluation = train_transformed @ evaluation.components_
print("Predict rating evaluation shape: ",predict_rating_evaluation.shape)
```

```python
known_test = test[test['Movie_id'].isin(train_matrix.columns)]
print("Test: ",test.shape[0])
print("Test for known movies: ",known_test.shape[0])
```

```python
known_test = known_test.copy()

known_test["predicted_rating"] = known_test.apply(
    lambda row: predict_rating_evaluation[
        train_matrix.index.get_loc(row["User_id"]),
        train_matrix.columns.get_loc(row["Movie_id"])
    ],
    axis=1
)

known_test[["User_id", "Movie_id", "rating", "predicted_rating"]].head()
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(
    mean_squared_error(
        known_test["rating"],
        known_test["predicted_rating"]
    )
)

print("RMSE:", rmse)
```

```python
k = 10

relevant_test = known_test[known_test["rating"] >= 4]

print("Relevant test ratings:", relevant_test.shape[0])
```

```python
k = 10

precision_scores = []

for user_id in train_matrix.index:
    user_index = train_matrix.index.get_loc(user_id)
    
    scores = predict_rating_evaluation[user_index]
    
    rated_movies = train_matrix.loc[user_id].dropna().index
    
    movie_ids = train_matrix.columns
    
    candidates = movie_ids[~movie_ids.isin(rated_movies)]
    
    candidate_scores = scores[train_matrix.columns.isin(candidates)]
    
    top_indices = candidate_scores.argsort()[::-1][:k]
    
    recommended_movies = candidates[top_indices]
    
    relevant_movies = relevant_test[
        relevant_test["User_id"] == user_id
    ]["Movie_id"]
    
    hits = len(set(recommended_movies) & set(relevant_movies))
    
    precision = hits / k
    
    precision_scores.append(precision)

precision_at_10 = np.mean(precision_scores)

print("Precision@10:", precision_at_10)
```

```python
recall_scores = []

for user_id in train_matrix.index:
    user_index = train_matrix.index.get_loc(user_id)
    
    scores = predict_rating_evaluation[user_index]
    
    rated_movies = train_matrix.loc[user_id].dropna().index
    
    movie_ids = train_matrix.columns
    candidates = movie_ids[~movie_ids.isin(rated_movies)]
    
    candidate_scores = scores[train_matrix.columns.isin(candidates)]
    
    top_indices = candidate_scores.argsort()[::-1][:10]
    
    recommended_movies = candidates[top_indices]
    
    relevant_movies = relevant_test[
        relevant_test["User_id"] == user_id
    ]["Movie_id"]
    
    if len(relevant_movies) > 0:
        hits = len(set(recommended_movies) & set(relevant_movies))
        recall = hits / len(relevant_movies)
        recall_scores.append(recall)

recall_at_10 = np.mean(recall_scores)

print("Recall@10:", recall_at_10)
```

```python
def recommend_svd(user_id, n=10):
    
    user_index = train_matrix.index.get_loc(user_id)
    
    user_scores = predict_rating_evaluation[user_index]
    
    rated_movies = train_matrix.loc[user_id].dropna().index
    
    movie_ids = train_matrix.columns
    
    candidate_movies = movie_ids[~movie_ids.isin(rated_movies)]
    
    candidate_scores = user_scores[train_matrix.columns.isin(candidate_movies)]
    
    top_indices = candidate_scores.argsort()[::-1][:n]
    
    recommended_movie_ids = candidate_movies[top_indices]
    
    recommended_movies = df1[
        df1["Movie_id"].isin(recommended_movie_ids)
    ]
    
    return recommended_movies[["Movie_id", "Title"]]
```
